# 🐍 Python pour la bio-analyse — Séance 1 : la condition `if`

**Objectif** : apprendre à faire prendre une décision à un programme.

En analyse de données (génomique, imagerie), on doit tout le temps décider :
*« cette lecture est-elle de bonne qualité ? », « ce pixel est-il assez lumineux ? »*
C'est le rôle de `if`.

## 1. Le principe

```python
if condition:
    # exécuté seulement si la condition est vraie
elif autre_condition:
    # exécuté si la première est fausse ET que celle-ci est vraie
else:
    # exécuté dans tous les autres cas
```

⚠️ Deux points à ne jamais oublier : le **`:`** en fin de ligne et l'**indentation** (4 espaces) du bloc.

Opérateurs de comparaison : `==` (égal), `!=` (différent), `>`, `<`, `>=`, `<=`

## 2. Exemple guidé (imagerie)
Un pixel d'image en niveaux de gris a une valeur entre 0 (noir) et 255 (blanc).

````python
pixel = 32

if pixel > 128:
    print("Pixel clair")
else:
    print("Pixel sombre")
````
